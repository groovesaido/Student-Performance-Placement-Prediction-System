```python
import opendatasets as od
import pandas as pd
```

```python
od.download('https://www.kaggle.com/datasets/shambhurajejagadale/student-performance-prediction-dataset')
```

```python
data=pd.read_csv('student_dataset_10000_rows.csv')
```

```python
data.info()
```

```python
from sklearn.model_selection import train_test_split
x=data.drop(columns=['exam_score','placement_status'])
y1=data['placement_status']
y2=data['exam_score']
x_train,x_test,y1_train,y1_test,y2_train,y2_test=train_test_split(x,y1,y2,test_size=0.2,random_state=42)
```

```python
#cleaning & preprocessing
from sklearn.preprocessing import OneHotEncoder

encoder=OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
).set_output(transform="pandas")

y1_train=encoder.fit_transform(y1_train.to_frame())
y1_test=encoder.transform(y1_test.to_frame())
```

```python
#scalling
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
#model regression
from xgboost import XGBRegressor

regmodel = XGBRegressor()
regmodel.fit(x_train, y2_train)
```

```python
#model classification
from sklearn.ensemble import RandomForestClassifier

clfmodel = RandomForestClassifier()
clfmodel.fit(x_train, y1_train)
```

```python
from sklearn.ensemble import RandomForestRegressor
regressor= RandomForestRegressor(n_estimators=400,random_state=42)
regressor.fit(x_train,y2_train)
```

```python
#regression score 2
round((regressor.score(x_test,y2_test)*100))
```

```python
#regression score 1
print(f"Accuracy: {regmodel.score(x_test, y2_test) * 100:.2f}%")
```

```python
#classification score
print(f"Accuracy: {clfmodel.score(x_test,y1_test)*100:.2f}")
```
